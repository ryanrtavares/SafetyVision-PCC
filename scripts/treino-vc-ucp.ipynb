```python
import torch
import shutil
import yaml
from ultralytics import YOLO
import os
import base64
import subprocess
import shutil
from IPython.display import FileLink
```

```python
## esse processo torna o treinamento muito mais rápido
print("Copiando 30.000 imagens para /kaggle/working/... Isso deve levar cerca de 1 a 2 minutos.")
caminho_origem = '/kaggle/input/datasets/businesssolutionsds/dataset-ppe/dataset'
caminho_destino = '/kaggle/working/dataset'

shutil.copytree(caminho_origem, caminho_destino, dirs_exist_ok=True)
print("Cópia concluída com sucesso!")
```

```python
## criando o yaml dinamicamente no kaggle para que todos os dados fiquem alocados em cache (pasta working)

caminho_base = '/kaggle/working/dataset'

print(f"Procurando dataset em: {caminho_base}\n")

train_path = os.path.join(caminho_base, 'images', 'train')
val_path = os.path.join(caminho_base, 'images', 'val')

print("Train:", train_path)
print("Existe:", os.path.exists(train_path))

print("Val:", val_path)
print("Existe:", os.path.exists(val_path))

if os.path.exists(train_path) and os.path.exists(val_path):

    data_yaml = {
        'path': caminho_base,
        'train': 'images/train',
        'val': 'images/val',

        'names': [
            'Fall-Detected',
            'Gloves',
            'Goggles',
            'Hardhat',
            'Ladder',
            'Mask',
            'NO-Gloves',
            'NO-Goggles',
            'NO-Hardhat',
            'NO-Mask',
            'NO-Safety Vest',
            'Person',
            'Safety Cone',
            'Safety Vest'
        ]
    }

    with open('/kaggle/working/data.yaml', 'w') as f:
        yaml.safe_dump(
            data_yaml,
            f,
            default_flow_style=False,
            sort_keys=False
        )

    print("\n✅ data.yaml criado corretamente!\n")
    print(open('/kaggle/working/data.yaml').read())

else:
    print("\n❌ Não encontrei as pastas de treino/validação.")
```

```python
model = YOLO('yolo11m.pt')

results = model.train(
    data='/kaggle/working/data.yaml',
    epochs=120,
    patience=30,
    imgsz=960,
    device=[0, 1],
    batch=16,
    workers=4,
    optimizer="AdamW",
    lr0=0.001,
    weight_decay=0.0005,
    cos_lr=True,
    close_mosaic=15,
    amp=True,
    seed=42
)
```

```python
def baixa_best_last(op, dir):
    """Passe best ou last como parâmetro para baixar o treinamento e a numeração do diretório na máquina kaggle"""
    if op.lower() == "last":
        shutil.copy('/kaggle/working/runs/detect/train-{dir}/weights/last.pt', '/kaggle/working/last.pt')
    elif op.lower() == "best":
        shutil.copy('/kaggle/working/runs/detect/train-{dir}/weights/best.pt', '/kaggle/working/best.pt')
```
